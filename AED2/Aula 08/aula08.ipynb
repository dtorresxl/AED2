```python
class Node:
    # Construtor
    def __init__(self, init_data):
        self.__data = init_data
        self.__next = None  # inicialmente não aponta para ninguém

    # Obtém o dado armazenado
    def get_data(self):
        return self.__data

    # Retorna o próximo elemento (para quem o nó aponta)
    def get_next(self):
        return self.__next

    # Armazena uma nova informação (atualiza dados)
    def set_data(self, new_data):
        self.__data = new_data

    # Aponta o nó para outro nó
    def set_next(self, new_next):
        self.__next = new_next


# Criando um nó e consultando o valor
temp = Node(93)
print(temp.get_data())
```

```python
class UnorderedList:
    # Construtor (aponta cabeça da lista para None)
    def __init__(self):
        self.head = None

    # Verifica se a lista está vazia
    def is_empty(self):
        return self.head == None

    # Adiciona elemento no início da lista - O(1)
    def add_head(self, item):
        # Cria novo nó
        temp = Node(item)
        # Aponta novo nó para a cabeça da lista
        temp.set_next(self.head)
        # Atualiza a cabeça da lista
        self.head = temp

    # Adiciona elemento no final da lista - O(n)
    def add_tail(self, item):
        # Cria novo nó
        tail = Node(item)
        # Se a lista está vazia, o novo nó é a cabeça
        if self.head == None:
            self.head = tail
            return
        # Usa referência temporária para percorrer a lista (aponta na cabeça)
        temp = self.head
        # Percorre a lista até o último elemento
        while temp.get_next() != None:
            temp = temp.get_next()
        # Aponta o último elemento para o novo nó
        temp.set_next(tail)

    # Imprime (retorna) os elementos da lista
    def print_list(self):
        temp = self.head
        X = []
        while temp != None:
            X.append(temp.get_data())
            temp = temp.get_next()
        return X

    # Calcula o número de elementos na lista - O(n)
    def size(self):
        # Aponta para cabeça da lista
        temp = self.head
        count = 0
        while temp != None:
            count = count + 1
            temp = temp.get_next()
        return count

    # Busca pelo elemento na lista - O(n)
    def search(self, item):
        # Inicia na cabeça da lista
        temp = self.head
        found = False
        # Percorre a lista até achar o elemento ou chegar no final
        while temp != None and not found:
            # Se o nó atual contém o elemento, found == True
            if temp.get_data() == item:
                found = True
            else:
                # Se não, aponta para o sucessor
                temp = temp.get_next()
        return found

    # Remove um nó da lista encadeada - O(n)
    def remove(self, item):
        # Aponta a referência corrente para a cabeça da lista
        current = self.head
        # Aponta a referência previous para None
        previous = None
        found = False
        # Enquanto não encontrar o valor a ser removido
        while not found and current != None:
            # Se o nó corrente armazena o item desejado, encontrou
            if current.get_data() == item:
                found = True
            else:
                # Atualiza o previous e o corrente
                previous = current
                current = current.get_next()
        # Se o item não existe na lista, não há o que remover
        if not found:
            return False
        # Se o nó a ser removido for o primeiro da lista, altera a cabeça
        if previous == None:
            self.head = current.get_next()
        else:
            # Caso contrário, liga o previous com o próximo
            previous.set_next(current.get_next())
        # Desliga o nó corrente
        current.set_next(None)
        return True
```

```python
class OrderedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head == None

    # Busca em lista ordenada: pode parar antes do fim
    def search(self, item):
        # Início da lista
        current = self.head
        found = False
        stop = False
        # Enquanto não atingir o final, não encontrou e não parou
        while current != None and not found and not stop:
            # Se o nó atual é o elemento, encontrou
            if current.get_data() == item:
                found = True
            else:
                # Senão, se o elemento atual é maior que o valor buscado, pare
                if current.get_data() > item:
                    stop = True
                else:
                    # Caso contrário, vai para o próximo
                    current = current.get_next()
        return found

    # Adiciona elemento na posição correta da lista
    def add(self, item):
        # Inicia na cabeça da lista
        current = self.head
        previous = None
        stop = False
        # Enquanto não chegar no final e não parou
        while current != None and not stop:
            # Se o valor do corrente for maior que o elemento desejado, pare
            if current.get_data() > item:
                stop = True
            else:
                # Senão, move o prévio e o corrente para o próximo
                previous = current
                current = current.get_next()
        # Cria novo nó
        temp = Node(item)
        # Se for o primeiro elemento, prévio = None
        if previous == None:
            temp.set_next(self.head)
            self.head = temp
        else:
            # Senão, estamos no meio ou no último
            temp.set_next(current)
            previous.set_next(temp)

    def print_list(self):
        temp = self.head
        X = []
        while temp != None:
            X.append(temp.get_data())
            temp = temp.get_next()
        return X


# Teste (os valores da figura do slide: 17, 26, 54, 77, 93 e inserindo 31)
O = OrderedList()
for v in [54, 26, 93, 17, 77]:
    O.add(v)
print(O.print_list())
O.add(31)
print(O.print_list())
print(O.search(31))
print(O.search(100))
```

```python
class Empty(Exception):
    pass


class DNode:
    # Construtor
    def __init__(self, init_data, prev, next):
        self.data = init_data
        self.prev = prev
        self.next = next

    # Obtém o dado armazenado
    def get_data(self):
        return self.data

    # Atualiza dado armazenado
    def set_data(self, new_data):
        self.data = new_data


class DoubleList:
    # Construtor
    def __init__(self):
        # Cria nós iniciais e finais (sentinelas)
        self.header = DNode(None, None, None)
        self.trailer = DNode(None, None, None)
        # header aponta para trailer e vice-versa
        self.header.next = self.trailer
        self.trailer.prev = self.header
        self.size = 0

    # Verifica se a lista está vazia
    def is_empty(self):
        return self.size == 0

    # Retorna o número de elementos na lista (função len)
    def __len__(self):
        return self.size

    # Insere novo nó entre dois nós existentes
    def insert_between(self, item, predecessor, successor):
        newest = DNode(item, predecessor, successor)
        predecessor.next = newest
        successor.prev = newest
        self.size += 1
        return newest

    # Remove um nó intermediário da lista
    # Header e trailer nunca podem ser removidos!
    def delete_node(self, node):
        predecessor = node.prev
        successor = node.next
        predecessor.next = successor
        successor.prev = predecessor
        self.size -= 1
        # Armazena o elemento removido
        element = node.data
        node.prev = node.next = node.data = None
        return element

    # Insere elemento no início - O(1)
    def insert_first(self, data):
        # nó deve entrar entre header e header.next
        self.insert_between(data, self.header, self.header.next)

    # Insere elemento no final - O(1)
    def insert_last(self, data):
        # nó deve entrar entre trailer.prev e trailer
        self.insert_between(data, self.trailer.prev, self.trailer)

    # Remove elemento no início - O(1)
    def delete_first(self):
        if self.is_empty():
            raise Empty('Lista está vazia!')
        return self.delete_node(self.header.next)

    # Remove elemento no final - O(1)
    def delete_last(self):
        if self.is_empty():
            raise Empty('Lista está vazia!')
        return self.delete_node(self.trailer.prev)

    # Imprime (retorna) elementos da lista
    def print_list(self):
        # Aponta referência para o primeiro nó real
        temp = self.header.next
        X = []
        # Percorre a lista adicionando elementos em X
        while temp.next != None:
            X.append(temp.data)
            temp = temp.next
        return X
```

```python
def get_length(linked_list):
    count = 0
    current = linked_list.head
    while current:
        count += 1
        current = current.get_next()
    return count


# Exemplo de uso
ll = UnorderedList()
ll.add_tail(1)
ll.add_tail(2)
ll.add_tail(3)
print(f"Comprimento da lista: {get_length(ll)}")
```

```python
class HistoricoNavegacao:
    def __init__(self):
        self.head = None

    # 1. Adicionar uma nova página ao histórico (no final)
    def adicionar_pagina(self, url):
        novo = Node(url)
        if self.head is None:
            self.head = novo
            return
        atual = self.head
        while atual.get_next() is not None:
            atual = atual.get_next()
        atual.set_next(novo)

    # 2. Exibir o histórico completo
    def exibir_historico(self):
        if self.head is None:
            print("Histórico vazio.")
            return
        atual = self.head
        paginas = []
        while atual is not None:
            paginas.append(atual.get_data())
            atual = atual.get_next()
        print(" -> ".join(paginas))

    # 3. Voltar uma página (remover o último elemento)
    def voltar(self):
        if self.head is None:
            print("Não há páginas para voltar.")
            return None
        # Só uma página: esvazia a lista
        if self.head.get_next() is None:
            removida = self.head.get_data()
            self.head = None
            return removida
        anterior = None
        atual = self.head
        while atual.get_next() is not None:
            anterior = atual
            atual = atual.get_next()
        anterior.set_next(None)
        return atual.get_data()


# Teste
h = HistoricoNavegacao()
h.adicionar_pagina("google.com")
h.adicionar_pagina("fatec.sp.gov.br")
h.adicionar_pagina("github.com")
h.exibir_historico()
print("Voltou de:", h.voltar())
h.exibir_historico()
h.voltar()
h.voltar()
h.voltar()
h.exibir_historico()
```

```python
class ListaEspera:
    def __init__(self):
        self.head = None
        self.tail = None   # guarda o último nó: inserir no final fica O(1)
        self.tamanho = 0

    # 1. Adicionar um paciente ao final da lista de espera
    def adicionar_paciente(self, nome):
        novo = Node(nome)
        if self.head is None:
            self.head = novo
            self.tail = novo
        else:
            self.tail.set_next(novo)
            self.tail = novo
        self.tamanho += 1

    # 2. Chamar o próximo paciente (remover o primeiro da lista)
    def chamar_proximo(self):
        if self.head is None:
            print("Ninguém na lista de espera.")
            return None
        chamado = self.head
        self.head = chamado.get_next()
        if self.head is None:
            self.tail = None
        chamado.set_next(None)
        self.tamanho -= 1
        return chamado.get_data()

    # 3. Exibir a lista de espera
    def exibir(self):
        if self.head is None:
            print("Lista de espera vazia.")
            return
        atual = self.head
        posicao = 1
        while atual is not None:
            print(f"{posicao}. {atual.get_data()}")
            atual = atual.get_next()
            posicao += 1


# Teste
espera = ListaEspera()
espera.adicionar_paciente("Ana")
espera.adicionar_paciente("Bruno")
espera.adicionar_paciente("Carla")
espera.exibir()
print("Chamando:", espera.chamar_proximo())
espera.exibir()
print("Chamando:", espera.chamar_proximo())
print("Chamando:", espera.chamar_proximo())
print("Chamando:", espera.chamar_proximo())
espera.exibir()
```
