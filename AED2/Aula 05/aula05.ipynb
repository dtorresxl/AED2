```python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def faz_aniversario(self):
        self.idade += 1

    def __str__(self):
        return f"{self.nome}, {self.idade} anos"


pessoa1 = Pessoa("Ana", 25)
pessoa2 = Pessoa("Bruno", 30)

print("Antes do aniversário:")
print(pessoa1)
print(pessoa2)

pessoa1.faz_aniversario()
pessoa2.faz_aniversario()

print("\nDepois do aniversário:")
print(pessoa1)
print(pessoa2)
```

```python
class Lampada:
    def __init__(self):
        self.__estado = False  # atributo privado

    def ligar(self):
        self.__estado = True

    def desligar(self):
        self.__estado = False

    def is_acesa(self):
        return self.__estado


lampada = Lampada()
print("Lâmpada acesa?", lampada.is_acesa())

lampada.ligar()
print("Após ligar() -> acesa?", lampada.is_acesa())

lampada.desligar()
print("Após desligar() -> acesa?", lampada.is_acesa())
```

```python
class Livro:
    def __init__(self, titulo, autor, quantidade_disponivel):
        self.titulo = titulo
        self.autor = autor
        self.__quantidade_disponivel = quantidade_disponivel  # protegido

    def esta_disponivel(self):
        return self.__quantidade_disponivel > 0

    def get_quantidade_disponivel(self):
        return self.__quantidade_disponivel

    def _emprestar(self):
        if self.__quantidade_disponivel > 0:
            self.__quantidade_disponivel -= 1
            return True
        return False

    def _devolver(self):
        self.__quantidade_disponivel += 1

    def __str__(self):
        return f'"{self.titulo}" ({self.autor}) - disponíveis: {self.__quantidade_disponivel}'


class Usuario:
    def __init__(self, nome, matricula):
        self.nome = nome
        self.matricula = matricula
        self.livros_emprestados = []

    def emprestar_livro(self, livro):
        if livro.esta_disponivel():
            sucesso = livro._emprestar()
            if sucesso:
                self.livros_emprestados.append(livro)
                print(f'{self.nome} pegou emprestado: "{livro.titulo}"')
                return True
        print(f'Empréstimo negado: "{livro.titulo}" está indisponível.')
        return False

    def devolver_livro(self, livro):
        if livro in self.livros_emprestados:
            livro._devolver()
            self.livros_emprestados.remove(livro)
            print(f'{self.nome} devolveu: "{livro.titulo}"')
        else:
            print(f'{self.nome} não possui o livro "{livro.titulo}" emprestado.')
```

```python
livro1 = Livro("Dom Casmurro", "Machado de Assis", 1)
livro2 = Livro("O Pequeno Príncipe", "Saint-Exupéry", 2)

ana = Usuario("Ana", "2024001")
bruno = Usuario("Bruno", "2024002")

ana.emprestar_livro(livro1)
bruno.emprestar_livro(livro1)  # deve falhar, só havia 1 exemplar
bruno.emprestar_livro(livro2)
ana.devolver_livro(livro1)
```

```python
class Conta:
    def __init__(self, titular, saldo=0):
        self.titular = titular
        self._saldo = saldo  # protegido, acessível por subclasses

    def depositar(self, valor):
        if valor > 0:
            self._saldo += valor
            print(f'{self.titular} depositou R$ {valor:.2f}. Saldo atual: R$ {self._saldo:.2f}')
        else:
            print("Valor de depósito inválido.")

    def sacar(self, valor):
        if valor > self._saldo:
            print(f'Saque de R$ {valor:.2f} negado: saldo insuficiente (saldo: R$ {self._saldo:.2f}).')
            return False
        self._saldo -= valor
        print(f'{self.titular} sacou R$ {valor:.2f}. Saldo atual: R$ {self._saldo:.2f}')
        return True

    def get_saldo(self):
        return self._saldo

    def __str__(self):
        return f'{self.__class__.__name__} de {self.titular} - Saldo: R$ {self._saldo:.2f}'


class ContaCorrente(Conta):
    taxa_saque = 2.00  # atributo de classe: taxa fixa por saque

    def sacar(self, valor):
        valor_total = valor + self.taxa_saque
        if valor_total > self._saldo:
            print(f'Saque negado: R$ {valor:.2f} + taxa de R$ {self.taxa_saque:.2f} excede o saldo.')
            return False
        self._saldo -= valor_total
        print(f'{self.titular} sacou R$ {valor:.2f} (+ taxa R$ {self.taxa_saque:.2f}). Saldo: R$ {self._saldo:.2f}')
        return True


class ContaPoupanca(Conta):
    juros = 0.05  # 5% ao ano, atributo de classe

    def render_juros(self):
        rendimento = self._saldo * self.juros
        self._saldo += rendimento
        print(f'{self.titular} rendeu R$ {rendimento:.2f} de juros. Saldo atual: R$ {self._saldo:.2f}')
```

```python
cc = ContaCorrente("Carlos", 500)
cc.depositar(200)
cc.sacar(100)
cc.sacar(1000)  # deve falhar

cp = ContaPoupanca("Marina", 1000)
cp.depositar(500)
cp.render_juros()
```
