```python
agenda = {}

def adicionar_contato(nome, telefone, email):
    agenda[nome] = {"telefone": telefone, "email": email}

def buscar_contato(nome):
    if nome in agenda:
        return agenda[nome]
    else:
        return "Contato não encontrado."

def listar_contatos():
    for nome, dados in agenda.items():
        print(f"{nome} - Telefone: {dados['telefone']} - E-mail: {dados['email']}")

# Testando
adicionar_contato("Ana", "1234-5678", "ana@email.com")
adicionar_contato("Bruno", "9876-5432", "bruno@email.com")

print("Buscando Ana:")
print(buscar_contato("Ana"))

print("\nListando todos os contatos:")
listar_contatos()
```

```python
tarefas = []

def adicionar_tarefa(descricao):
    tarefas.append({"descricao": descricao, "concluida": False})

def concluir_tarefa(descricao):
    for tarefa in tarefas:
        if tarefa["descricao"] == descricao:
            tarefa["concluida"] = True
            return
    print("Tarefa não encontrada.")

def listar_pendentes():
    pendentes = [t["descricao"] for t in tarefas if not t["concluida"]]
    print("Tarefas pendentes:", pendentes)

# Testando
adicionar_tarefa("Estudar Python")
adicionar_tarefa("Fazer compras")
adicionar_tarefa("Lavar o carro")

concluir_tarefa("Fazer compras")

listar_pendentes()
```
