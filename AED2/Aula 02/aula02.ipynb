```python
def imprime_primeiro(lista):
    print(lista[0])

#Complexidade: O(1) — tempo constante. Acessar um elemento por índice não depende do tamanho da lista.
```

```python
def encontrar_maximo(lista):
    valor_maximo = lista[0]
    for numero in lista:
        if numero > valor_maximo:
            valor_maximo = numero
    return valor_maximo

#Complexidade: O(n) — linear. Percorre todos os n elementos uma vez.
```

```python
def busca_binaria(lista, alvo):
    esquerda = 0
    direita = len(lista) - 1
    while esquerda <= direita:
        meio = (esquerda + direita) // 2
        if lista[meio] == alvo:
            return meio
        elif lista[meio] < alvo:
            esquerda = meio + 1
        else:
            direita = meio - 1
    return -1

#Complexidade: O(log n) — o espaço de busca é dividido pela metade a cada iteração.
```

```python
def existe_2_menores_v1(idades):
    tamanho = len(idades)
    menor = 200
    for i in range(tamanho):
        if idades[i] < menor:
            menor = idades[i]
    cont = 0
    for i in range(tamanho):
        if idades[i] == menor:
            cont += 1
    return cont > 1

def existe_2_menores_v2(idades):
    idades.sort()
    return idades[0] == idades[1]


#v1: dois laços simples → O(n)
#v2: usa sort() → O(n log n)
#A v1 é mais eficiente, pois O(n) cresce mais devagar que O(n log n).
```

```python
def fib(n):
    if n < 2:
        return n
    else:
        return fib(n-1) + fib(n-2)
```

```python
ex_alunos = [
    {"nome": "Ana Silva", "cpf": "111.111.111-11", "curso": "ADS"},
    {"nome": "Bruno Souza", "cpf": "222.222.222-22", "curso": "Gestão Empresarial"},
    {"nome": "Carla Mendes", "cpf": "333.333.333-33", "curso": "ADS"},
    {"nome": "Daniel Costa", "cpf": "444.444.444-44", "curso": "Logística"},
]

def buscar_aluno_linear(lista_alunos, cpf_buscado):
    for aluno in lista_alunos:
        if aluno["cpf"] == cpf_buscado:
            return aluno
    return None

def criar_indice_por_cpf(lista_alunos):
    return {aluno["cpf"]: aluno for aluno in lista_alunos}

def buscar_aluno_eficiente(indice_alunos, cpf_buscado):
    return indice_alunos.get(cpf_buscado)

cpf_alvo = "333.333.333-33"
print(buscar_aluno_linear(ex_alunos, cpf_alvo))
print(buscar_aluno_eficiente(criar_indice_por_cpf(ex_alunos), cpf_alvo))
```

```python
def recomendar_disciplinas(historico_aluno, historicos_outros_alunos):
    disciplinas_aluno = set(historico_aluno)
    recomendacoes = set()
    for historico in historicos_outros_alunos:
        disciplinas_outro = set(historico)
        if disciplinas_aluno & disciplinas_outro:
            novas = disciplinas_outro - disciplinas_aluno
            recomendacoes.update(novas)
    return list(recomendacoes)

historico_ana = ["MAT101", "PROG101", "BD101"]
historicos_outros = [
    ["PROG101", "REDES101", "ENG201"],
    ["QUIM101", "FIS101"],
    ["BD101", "SIST_OPER101", "MAT101"],
]

print(recomendar_disciplinas(historico_ana, historicos_outros))
```
