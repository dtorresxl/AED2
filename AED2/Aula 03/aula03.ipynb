```python
def buscar_cliente(nome, lista_clientes):
    for i, cliente in enumerate(lista_clientes):
        if cliente == nome:
            return i
    return -1

lista_clientes = ["Ana", "Bruno", "Carla", "Daniel", "Eduarda"]

nome_existente = "Carla"
nome_inexistente = "Fernando"

indice1 = buscar_cliente(nome_existente, lista_clientes)
if indice1 != -1:
    print(f'Cliente "{nome_existente}" encontrado no índice {indice1}.')
else:
    print(f'Cliente "{nome_existente}" não encontrado.')

indice2 = buscar_cliente(nome_inexistente, lista_clientes)
if indice2 != -1:
    print(f'Cliente "{nome_inexistente}" encontrado no índice {indice2}.')
else:
    print(f'Cliente "{nome_inexistente}" não encontrado.')
```

```python
def conta_ocorrencias(valor, lista):
    contador = 0
    for elemento in lista:
        if elemento == valor:
            contador += 1
    return contador

print(conta_ocorrencias(3, [1, 3, 5, 3, 7]))
print(conta_ocorrencias(9, [9, 9, 9, 9]))
print(conta_ocorrencias(2, [1, 3, 5, 7]))
print(conta_ocorrencias("banana", ["maçã", "banana", "banana", "uva"]))
```

```python
def remove_elemento(lista, valor):
    for i in range(len(lista)):
        if lista[i] == valor:
            return lista.pop(i)
    return None

numeros = [10, 20, 30, 40, 50]
print("Lista original:", numeros)

removido = remove_elemento(numeros, 30)
print("Valor removido:", removido)
print("Lista após remoção:", numeros)

removido2 = remove_elemento(numeros, 999)
print("Tentando remover 999 (não existe):", removido2)
```

```python
def buscar_filme(titulo, lista_filmes):
    for i, filme in enumerate(lista_filmes):
        if filme.lower() == titulo.lower():
            return i
    return -1

filmes = ["Matrix", "Interestelar", "Cidade de Deus", "O Poderoso Chefão", "Parasita"]

titulo_busca1 = "Interestelar"
indice = buscar_filme(titulo_busca1, filmes)
if indice != -1:
    print(f'Filme "{titulo_busca1}" encontrado na posição {indice} do catálogo.')
else:
    print(f'Filme "{titulo_busca1}" não está no catálogo.')

titulo_busca2 = "Vingadores"
indice = buscar_filme(titulo_busca2, filmes)
if indice != -1:
    print(f'Filme "{titulo_busca2}" encontrado na posição {indice} do catálogo.')
else:
    print(f'Filme "{titulo_busca2}" não está no catálogo.')
```

```python
def verificar_convidado(nome, lista_convidados):
    for convidado in lista_convidados:
        if convidado.lower() == nome.lower():
            return True
    return False

convidados = ["Marcos", "Juliana", "Rafael", "Beatriz", "Tiago"]

nome_busca1 = "Beatriz"
if verificar_convidado(nome_busca1, convidados):
    print(f'"{nome_busca1}" está na lista de convidados.')
else:
    print(f'"{nome_busca1}" NÃO está na lista de convidados.')

nome_busca2 = "Larissa"
if verificar_convidado(nome_busca2, convidados):
    print(f'"{nome_busca2}" está na lista de convidados.')
else:
    print(f'"{nome_busca2}" NÃO está na lista de convidados.')
```
