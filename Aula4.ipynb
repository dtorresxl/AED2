```python
livros = [
    {"A Menina que Roubava Livros", "Markus Zusak", 2005},
    {"Dom Casmurro", "Machado de Assis", 1899},
    {"Ensaio Sobre a Cegueira", "José Saramago", 1995},
    {"O Cortiço",  "Aluísio Azevedo", 1890},
    {"O Senhor dos Anéis",  "J. R. R. Tolkien", 1954},
    {"O Sol é Para Todos", "Harper Lee", 1960},
]
```

```python
def busca_binaria(livros, titulo):
    inicio = 0
    fim = len(livros) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2
        if livros[meio][0].lower() == titulo.lower():
            return meio
        elif livros[meio][0].lower() < titulo.lower():
            inicio = meio + 1
        else:
            fim = meio - 1

    return -1
```

```python
busca_titulo = input("Digite o título do livro que deseja buscar: ")
livro_encontrado = busca_binaria(livros, busca_titulo)

if livro_encontrado:
    print(f"Livro encontrado: {livros[livro_encontrado]}")
    print(f"Posição na lista: {livro_encontrado}")
    print(f"Total de livros na lista: {len(livros)}")
else:
    print("Livro não encontrado.")
```
